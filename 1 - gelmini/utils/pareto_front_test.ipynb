```python
from typing import List, Dict, Any

def get_pareto_front_2d(data: List[Dict[str, Any]], kpi1_key: str, kpi2_key: str) -> List[Dict[str, Any]]:
    """
    Extracts the Pareto front from a list of dictionaries based on 2 continuous variables (KPIs).
    Assumption: Both KPIs are subject to MINIMIZATION.
    
    Args:
        data: A list of dictionaries. Each dict represents a candidate solution.
        kpi1_key: The dictionary key corresponding to the first objective.
        kpi2_key: The dictionary key corresponding to the second objective.
        
    Returns:
        A new list of dictionaries containing only the non-dominated solutions.
    """
    
    # Edge case: empty list
    if not data:
        return []

    # Step 1: Sort the candidate solutions
    # Sort primarily by KPI 1 (ascending). 
    # In case of ties on KPI 1, sort by KPI 2 (ascending).
    sorted_data = sorted(data, key=lambda x: (x[kpi1_key], x[kpi2_key]))
    
    pareto_front = []
    min_kpi2_seen = float('inf')
    
    # Step 2: Linear scan to filter dominated solutions
    for current_point in sorted_data:
        current_kpi2 = current_point[kpi2_key]
        
        # Since 'sorted_data' is ordered by KPI 1, the current point is guaranteed 
        # to be worse (or equal) on KPI 1 compared to previously evaluated points.
        # Therefore, to be considered non-dominated and part of the Pareto front, 
        # it MUST be strictly better (<) on KPI 2.
        if current_kpi2 < min_kpi2_seen:
            pareto_front.append(current_point)
            # Update the bounding value for the next iterations
            min_kpi2_seen = current_kpi2
            
    return pareto_front

```
